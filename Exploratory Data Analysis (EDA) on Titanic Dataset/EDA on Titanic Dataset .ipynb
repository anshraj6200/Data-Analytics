<h3>Importing necessary libraries</h3>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

<h3>Setting plot styles for aesthetics</h3>

```python
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.rename(columns={'Sex': 'Gender', 'Pclass': 'Class'}, inplace=True)
```

```python
print("Shape of dataset:", df.shape)
print("\nDataset Info:")
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())
```

<h3>Survival Countplot</h3>

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df, palette='pastel')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.title('Survival Count')
plt.xlabel('Survival')
plt.ylabel('Passenger Count')
plt.show()
```

<h3>Survival by Gender and Class</h3>

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Class', hue='Gender', data=df[df['Survived'] == 1], palette='Set2')
plt.title('Survived Passengers by Gender and Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count of Survivors')
plt.legend(title='Gender')
plt.show()
```

<h3>Pie Chart - Class Distribution</h3>

```python
class_counts = df['Class'].value_counts()
labels = ['Class 1', 'Class 2', 'Class 3']
plt.figure(figsize=(6, 6))
plt.pie(class_counts, labels=labels, autopct='%1.1f%%', startangle=140,
        colors=sns.color_palette("pastel"))
plt.title('Passenger Class Distribution')
plt.show()
```

<h3>Age Distribution Histogram</h3>

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'].dropna(), bins=30, kde=True, color='skyblue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```
